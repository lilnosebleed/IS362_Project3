```python
## chinook and sql database
## RDBMS Used: SQLite
import sqlite3
import pandas as pd


## establishs connection using the built in sqlite3
conn = sqlite3.connect("chinook.sqlite")

## sql query joins 5 tables: Customer - Invoice - InvoiceLine - Track - Album
## sorting by LastName, then FirstName
sql_query = """
SELECT
    c.LastName,
    c.FirstName,
    t.Name,
    a.Title
FROM Customer c
JOIN Invoice i ON c.CustomerId = i.CustomerId
JOIN InvoiceLine il ON i.InvoiceId = il.InvoiceId
JOIN Track t ON il.TrackId = t.TrackId
JOIN Album a ON t.AlbumId = a.AlbumId
ORDER BY c.LastName, c.FirstName
"""

## reads the sql query directly into a pandas DataFrame
df = pd.read_sql_query(sql_query, conn)

## displays the first 5 rows
df.head()
```

```python
## closing the database connection when finished
conn.close()
```
